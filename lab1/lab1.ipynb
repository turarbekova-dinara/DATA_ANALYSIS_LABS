```python
# Pandas кітапханасын қосу
import pandas as pd
```

```python
# CSV файлды оқу
df = pd.read_csv("StudentsPerformance.csv")
```

```python
# Алғашқы 5 жолды көрсету
df.head()
```

```python
#Сандар 1-ге жақын болса → байланыс күшті.
df.corr(numeric_only=True)
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["reading score"], df["writing score"])
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.title("Relationship between Reading and Writing Scores")
plt.show()
```

```python
"""
The analysis shows a strong positive relationship between reading and writing scores.
This suggests that students who perform well in reading tend to also perform well in writing.
Improving reading skills may therefore help improve writing performance.
"""
```
